## Reading Iceberg tables directly with Python (no Spark)

`pyiceberg` talks to the same REST catalog (`http://rest:8181`) that Spark and the
Flink pipeline use, so it sees the same tables. Useful when you want a plain
pandas DataFrame without going through a Spark session/SQL cell.

```python
from pyiceberg.catalog import load_catalog

catalog = load_catalog(
    "demo",
    **{
        "type": "rest",
        "uri": "http://rest:8181",
        "s3.endpoint": "http://minio:9000",
        "s3.access-key-id": "admin",
        "s3.secret-access-key": "password",
        "s3.path-style-access": "true",
    },
)

catalog.list_namespaces()
```

```python
table = catalog.load_table("payroll_dev.validated_employees")
print(table.schema())

df = table.scan().to_pandas()
df.head()
```
